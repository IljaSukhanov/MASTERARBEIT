# Local bandwidth minimizers across sample sizes

Compare $\widehat Q_0$, $\widehat Q_{13/14}$, $\widehat Q_1$, and an ISE reference using the same
1,000 paired standard-normal samples for every $n=100,200,\ldots,2000$.

For each sample and energy criterion, select the lowest-energy detected local
minimum strictly inside $(0.01,2)$. Samples without a minimum retain a
missing outcome; endpoints are never substituted. Histogram frequencies
divide by all 1,000 samples. Every candidate and its stationarity and
curvature diagnostics are saved separately.

The validated Q1 results for four sample sizes and the exact $n=200$
candidates are reused. New selections use Fourier derivative splines to
locate basins and exact Gaussian-pair derivatives to refine accepted roots.
Ambiguous basins use an exact curvature search. Validation doubles the scan
and spatial resolution and checks an independent exact derivative scan at
every sample size. These numerical checks do not certify exhaustive
stationary-point enumeration.

Sources: [energy implementation](../kde_tools/continuous_minimizers.py),
[search engine](../outputs/analysis/local_qr_minimizers_vs_n/local_search.py),
[workflow](../outputs/analysis/local_qr_minimizers_vs_n/run_experiment.py),
and [paired-sample settings](../outputs/data/continuous_ise_qhat_distributions_vs_n/thesis/5b5fa0b0a92f/settings.json).

The violet ISE histogram uses the bandwidth numerically minimizing ISE over
[0.01,2] against the known standard-normal density. Its original validated
histogram counts are recovered from the full-interval vector PDFs, using the
same 201 geometric bins and the same paired samples. Every recovered bin is
checked against the 4,000 saved individual ISE bandwidths at the four selected
sample sizes. The three local-minimum datasets are preserved.
