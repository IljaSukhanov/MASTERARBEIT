# Interior local minimizers of $\widehat Q_1$ on $(0.01,2)$

Use the same 1,000 paired standard-normal samples as Figure 3.9 at each of
$n=100,1000,1500,2000$.

For each sample, find local minima strictly inside $(0.01,2)$. If several
exist, use the one with the lowest energy. If none is found, retain a missing
outcome; never replace it with an endpoint. Histogram frequencies divide by
all 1,000 repetitions. Every detected candidate is retained separately.

A Fourier scan supplies gradient and curvature brackets. Exact Gaussian-pair
derivatives refine the candidates. Validation checks doubled scan and spatial
resolution and fresh exact derivative scans, including $n=200$ checks. This numerical procedure does not certify that
every possible local minimum has been found.

Sources: [energy and exact derivatives](../kde_tools/continuous_minimizers.py),
[full workflow](../outputs/analysis/local_q1_minimizers_selected_n/run_experiment.py),
and [original paired-sample settings](../inputs/qhat_sampling.json).
