# Fast L² bandwidth distributions: sample size and lambda

Start with the Figure 3.6 baseline: **ten separate bandwidth distributions** for $n=100,200,\ldots,1000$, at **$\lambda=1$**. The baseline and separate-lambda figures show **only spectrally chosen bandwidths**. The final saved-run comparison also overlays the sampling distribution of the continuous $\widehat Q_{n,2}$ minimizer.

Use the existing direct Fourier-quadrature L² implementation with

$$D_{ij}=\frac12\|\widehat f_{h_i}-\widehat f_{h_j}\|_2^2,\qquad S_{ij}=\exp(-\lambda D_{ij}).$$

Select the bandwidth at the largest Perron-vector component. Keep the original L² experiment's geometric grid, sampling seed, and nested samples. The default thesis profile uses $M=1000$ repetitions and $N=201$ candidates on $[0.01,2]$, with the periodic FFT interval $[-12,12)$ and 8,192 spatial points.

**Run the cells in order.** `LAMBDA_VALUES = [1.0]` gives the requested baseline first. To compare larger lambda values later, extend this list, for example to `[1.0, 10.0, 100.0, 1000.0]`, and rerun from the settings cell. Each lambda receives its own ten-panel figure. All lambdas share the same samples, and the Fourier-quadrature divergence matrix is calculated once per sample and reused across lambdas. Checkpoints and exports are kept separately for each complete configuration.

Sources: [Figure 3.6 / increasing-sample-size experiment, `fig:distribution_of_chosen_bandwidths_for_increasing_sample_size`](../../../Abschlussarbeit_2021/LaTeX/chapters/5_bandwidth_selection.tex); [original fast L² experiment, section 3 / `beta2_n_sweep_*`](../xx_notebooks_overview/spectral_scale_selection_kde.ipynb); [shared Fourier-quadrature divergence helper](../kde_tools/fourier/similarities.py); [shared periodic KDE reconstruction](../kde_tools/fourier/reconstruction.py).


L² inner products use direct frequency quadrature (`frequency_trapezoid_v1`) with the shared production settings **Omega = 450 and K = 1536 intervals (1537 nodes)**. The compatibility values `half_width = pi*K/Omega` and `spatial_points = 2*K` specify this frequency grid; the L² helper does not reconstruct spatial densities. Configuration identities include the frequency grid, so checkpoints from other grids are not reused. Rerun setup after updating the helpers.

## Settings: lambda = 1 baseline

The default values match the fast L² sample-size experiment: $n=100,200,\ldots,1000$, $M=1000$, $N=201$, $h_{\min}=0.01$, $h_{\max}=2$, and seed `20260830` (the original notebook's `seed + 10 + 2`). One maximum-size Gaussian sample is drawn per repetition and its prefixes are used at every $n$.

The preview profile reduces the repetition count to 100 while keeping the same bandwidth and FFT grids. `LAMBDA_VALUES` controls subsequent comparisons; leave `[1.0]` for the first experiment.

Source: [the original fast L² notebook's setup, beta-2 settings, and increasing-sample-size cell](../xx_notebooks_overview/spectral_scale_selection_kde.ipynb).


```python
RUN_PROFILE = "thesis"  # "thesis" or "preview"
PROFILES = {
    "thesis": {"repetitions": 1000, "grid_size": 201},
    "preview": {"repetitions": 100, "grid_size": 201},
}
SAMPLE_SIZES = list(range(100, 1001, 100))
LAMBDA_VALUES = [1.0]  # Start here; later extend to [1.0, 10.0, 100.0, 1000.0].
REUSE_CHECKPOINTS = True
VALIDATE_FFT = True

settings = {
    **PROFILES[RUN_PROFILE],
    "sigma2": 1.0,
    "h_min": 0.01,
    "h_max": 2.0,
    "frequency_cutoff": DEFAULT_FREQUENCY_CUTOFF,
    "frequency_intervals": DEFAULT_FREQUENCY_INTERVALS,
    "fft_half_width": DEFAULT_FOURIER_HALF_WIDTH,
    "fft_spatial_points": DEFAULT_FOURIER_SPATIAL_POINTS,
    "seed": 20260830,
    "maximum_sample_size": max(SAMPLE_SIZES),
}
LAMBDA_VALUES = [float(value) for value in LAMBDA_VALUES]
if (not LAMBDA_VALUES or LAMBDA_VALUES[0] != 1.0
        or not np.all(np.isfinite(LAMBDA_VALUES))
        or np.any(np.diff(LAMBDA_VALUES) <= 0)):
    raise ValueError("Start LAMBDA_VALUES with 1.0 and use finite, strictly increasing values.")
if settings["repetitions"] < 1 or settings["grid_size"] < 2:
    raise ValueError("Repetitions must be positive and the bandwidth grid needs two points.")
if settings["sigma2"] <= 0 or settings["fft_half_width"] <= 0:
    raise ValueError("Variance and FFT half-width must be positive.")
if settings["fft_spatial_points"] < 2 or settings["fft_spatial_points"] % 2:
    raise ValueError("The FFT grid needs an even number of at least two points.")
h_grid = bandwidth_grid_linear_in_psi_beta(
    h_min=settings["h_min"], h_max=settings["h_max"], beta=2.0,
    grid_size=settings["grid_size"],
)
configuration = {
    "settings": settings, "sample_sizes": SAMPLE_SIZES,
    "lambda_values": LAMBDA_VALUES, "method": "frequency_quadrature_l2_spectral",
    "implementation_version": 3, "spectral_sweep_runner": SPECTRAL_SWEEP_RUNNER_VERSION,
    "l2_method": FOURIER_L2_METHOD_VERSION, "perron_solver": PERRON_SOLVER_VERSION,
}
configuration_id = hashlib.sha256(json.dumps(configuration, sort_keys=True).encode()).hexdigest()[:12]
run_directory = data_output_directory / RUN_PROFILE / configuration_id
image_directory = image_output_directory / RUN_PROFILE / configuration_id
for directory in (run_directory, image_directory):
    directory.mkdir(parents=True, exist_ok=True)
(run_directory / "settings.json").write_text(
    json.dumps({"profile": RUN_PROFILE, **configuration}, indent=2) + "\n"
)
full_samples = np.random.default_rng(settings["seed"]).normal(
    0.0, np.sqrt(settings["sigma2"]),
    size=(settings["repetitions"], settings["maximum_sample_size"]),
)
print(f"Sample sizes: {SAMPLE_SIZES}; lambdas: {LAMBDA_VALUES}")
print(f"Profile: {RUN_PROFILE}; M={settings['repetitions']}; N={len(h_grid)}; "
      f"Omega={settings['frequency_cutoff']}, K={settings['frequency_intervals']}")
print(f"Checkpoints and CSV data: {run_directory}")
spectral_sweep_cache_root = code_directory / "outputs/data/shared_spectral_sweeps"
spectral_sweep_import_directories = [
    directory
    for profile in ("thesis", "preview")
    for directory in (data_output_directory / profile).glob("*")
    if directory.is_dir()
]
```

## Shared L² spectral selection

Each sample's Fourier-quadrature divergence matrix is reused for every configured lambda. The selection calls `spectral_choice_from_divergence` from `kde_tools.similarities`, which delegates to the centralized stabilized `eigh` solver in `kde_tools.perron`. It shifts the off-diagonal exponents before exponentiation and checks the eigenpair residual and spectral gap. Checkpoint configurations include the solver version.

For symmetric similarities, the Doob stationary mass is proportional to the square of the Perron vector and therefore has the same maximizer. See [`kde_tools/perron.py`](../kde_tools/perron.py) and the [SciPy `eigh` documentation](https://docs.scipy.org/doc/scipy/reference/generated/scipy.linalg.eigh.html).

The runner is shared with [section 3 of the main notebook](../xx_notebooks_overview/spectral_scale_selection_kde.ipynb).
A common cache under `outputs/data/shared_spectral_sweeps/` identifies the actual
sample prefixes, grid, quadrature settings, and solver versions. Completed
spectral choices can be reused across notebooks and lambda subsets; each new
lambda is computed only where it is missing. The existing per-run checkpoints
and CSV exports remain available. New configuration identifiers distinguish
this runner, and compatible earlier checkpoints are imported read-only.


```python
# Shared with section 3 of the main spectral KDE notebook.
# Imports are repeated here so this cell can be rerun after reloading helpers.
from kde_tools.spectral_sweeps import run_l2_point, validate_l2_checkpoint
```

## Small Fourier-quadrature validation

Compare the Fourier-quadrature divergence with the shared exact Gaussian formula on a small subset of bandwidths from the first sample. This check precedes the repeated simulation and does not compute any additional bandwidth-selection distribution.

Sources: [Fourier-quadrature validation in the original L² notebook](../xx_notebooks_overview/spectral_scale_selection_kde.ipynb); [exact Gaussian L² helper](../kde_tools/similarities.py).


```python
if VALIDATE_FFT:
    validation_indices = np.unique(np.linspace(0, len(h_grid) - 1, min(len(h_grid), 20), dtype=int))
    validation_h_grid = h_grid[validation_indices]
    validation_samples = full_samples[0, :min(100, settings["maximum_sample_size"])]
    validation_exact = beta2_divergence_matrix(validation_samples, validation_h_grid)
    validation_fft = beta2_divergence_matrix_fourier(
        validation_samples, validation_h_grid,
        half_width=settings["fft_half_width"], spatial_points=settings["fft_spatial_points"],
    )
    validation_absolute_error = float(np.max(np.abs(validation_fft - validation_exact)))
    validation_relative_error = validation_absolute_error / max(float(np.max(validation_exact)), np.finfo(float).tiny)
    print(f"FFT validation: maximum absolute D2 error={validation_absolute_error:.3e}; "
          f"error relative to max(D2)={validation_relative_error:.3e}")
    if validation_relative_error > 2e-5:
        raise ArithmeticError("FFT validation failed; increase the FFT resolution or interval before running.")
```

## Repeated spectral selections

The next cell computes the ten sample-size distributions for every configured lambda, sharing samples across both sample sizes and lambda values. A checkpoint is saved after every completed repetition. Rerunning with unchanged settings resumes that run.


```python
results_by_n = {}
for current_n in SAMPLE_SIZES:
    results_by_n[current_n] = run_l2_point(
        current_n, full_samples, h_grid, settings, LAMBDA_VALUES, run_directory,
        reuse=REUSE_CHECKPOINTS, cache_root=spectral_sweep_cache_root,
        import_directories=spectral_sweep_import_directories,
    )
print(f"Completed {len(SAMPLE_SIZES)} distributions per lambda for {LAMBDA_VALUES}.")
```

## Saved spectral selections and distributions

Export the chosen bandwidth for each repetition, the grid frequencies, and one summary row for each sample size and lambda. These files contain spectral selections only.


```python
selection_rows = []
frequency_rows = []
summary_rows = []
for current_n in SAMPLE_SIZES:
    result = results_by_n[current_n]
    for column, lambda_value in enumerate(LAMBDA_VALUES):
        chosen_indices = result["selected_indices"][:, column]
        chosen_h = h_grid[chosen_indices]
        selection_rows.extend(
            {"n": current_n, "lambda": lambda_value, "repetition": repetition,
             "grid_index": int(index), "h": float(h_grid[index])}
            for repetition, index in enumerate(chosen_indices)
        )
        frequency_rows.extend(
            {"n": current_n, "lambda": lambda_value, "grid_index": index,
             "h": float(h), "count": int(result["counts"][column, index]),
             "relative_frequency": float(result["frequencies"][column, index])}
            for index, h in enumerate(h_grid)
        )
        modal_index = int(np.argmax(result["counts"][column]))
        summary_rows.append({
            "n": current_n, "lambda": lambda_value, "repetitions": settings["repetitions"],
            "modal_h": float(h_grid[modal_index]), "mean_h": float(np.mean(chosen_h)),
            "median_h": float(np.median(chosen_h)),
            "modal_frequency": float(result["frequencies"][column, modal_index]),
        })
for filename, rows in (
    ("selections.csv", selection_rows), ("frequencies.csv", frequency_rows),
    ("summary.csv", summary_rows),
):
    with (run_directory / filename).open("w", newline="") as stream:
        writer = csv.DictWriter(stream, fieldnames=rows[0].keys())
        writer.writeheader()
        writer.writerows(rows)
print(f"CSV data: {run_directory}")
print(f"{'lambda':>9s} {'n':>5s} {'modal h':>10s} {'mean h':>10s} {'median h':>10s}")
for row in sorted(summary_rows, key=lambda row: (row["lambda"], row["n"])):
    print(f"{row['lambda']:9g} {row['n']:5d} {row['modal_h']:10.5f} "
          f"{row['mean_h']:10.5f} {row['median_h']:10.5f}")
```

## First figure: ten spectral L² distributions at lambda = 1

The first displayed figure always contains the ten distributions at $\lambda=1$, arranged in a 5 × 2 grid. Each panel contains one spectral histogram. Subsequent lambda values, when configured, receive the same layout in increasing order. Horizontal and vertical limits are shared across all displayed figures.

Every displayed figure is automatically saved as a PDF under `outputs/pdfs/fast_l2_bandwidth_distributions_vs_n_lambda/`; a PNG copy is also saved. No ISE distribution, Q-hat distribution, or reference line is added.

Source for the experiment and geometric-grid plotting convention: [original Figure 3.6 workflow](../xx_notebooks_overview/spectral_scale_selection_kde.ipynb).


```python
if set(results_by_n) != set(SAMPLE_SIZES):
    raise RuntimeError("Finish all ten sample sizes before plotting.")
grid_positions = np.arange(len(h_grid))
ticks = np.unique(np.linspace(0, len(h_grid) - 1, 7, dtype=int))
y_max = 1.08 * max(np.max(result["frequencies"]) for result in results_by_n.values())

for column, lambda_value in enumerate(LAMBDA_VALUES):
    fig, axes = plt.subplots(5, 2, figsize=(14, 18), sharex=True, sharey=True)
    for ax, current_n in zip(axes.flat, SAMPLE_SIZES):
        ax.bar(
            grid_positions, results_by_n[current_n]["frequencies"][column],
            width=0.86, color="#D6843B", alpha=0.85, edgecolor="white", linewidth=0.25,
        )
        ax.set_title(rf"$n={current_n}$", fontsize=13, pad=8)
        ax.set_xticks(ticks, [f"{h_grid[index]:.3g}" for index in ticks])
        ax.set_xlim(-0.7, len(h_grid) - 0.3)
        ax.set_ylim(0, y_max)
        ax.yaxis.set_major_formatter(PercentFormatter(xmax=1.0, decimals=0))
        ax.set_xlabel("spectrally selected bandwidth h", fontsize=10)
        ax.set_ylabel("relative frequency", fontsize=10)
        ax.tick_params(labelsize=9, colors="#304967", labelbottom=True, labelleft=True)
        ax.grid(axis="y", color="#DDE5EC", linewidth=0.7)
        ax.set_axisbelow(True)
        for side in ("top", "right"):
            ax.spines[side].set_visible(False)
        for side in ("left", "bottom"):
            ax.spines[side].set_color("#B8C5D1")
    fig.suptitle(
        rf"Fast $L^2$ spectral bandwidth distributions: $\lambda={lambda_value:g}$" + "\n"
        + rf"Gaussian data; $M={settings['repetitions']}$ per $n$, $N={len(h_grid)}$, "
        + r"$S=\exp(-\lambda D_2)$" + f" ({RUN_PROFILE})",
        fontsize=16, y=0.995,
    )
    fig.tight_layout(rect=(0, 0, 1, 0.965), h_pad=2.0, w_pad=2.0)
    lambda_tag = f"{lambda_value:g}".replace(".", "p").replace("+", "")
    png_path = image_directory / (
        f"spectral_l2_lambda{lambda_tag}_M{settings['repetitions']}_N{len(h_grid)}.png"
    )
    fig.savefig(png_path, dpi=160, bbox_inches="tight")
    plt.show()  # The shared automatic exporter saves each ten-panel figure as a PDF.
    plt.close(fig)
    print(f"PNG (lambda={lambda_value:g}): {png_path}")
print(f"PDF directory: {pdf_output_directory}")
```

## Comparison: four lambda values and continuous Q-hat minimizers

This figure has **ten panels**, one for each $n=100,200,\ldots,1000$. Each panel overlays the four spectral bandwidth distributions for $\lambda=1,10,100,1000$ and, in dashed black, the sampling distribution of

$$\widehat h_Q=\operatorname*{arg\,min}_{h\in[0.01,2]}\widehat Q_{n,2}(h),
\qquad\widehat Q_{n,2}(h)=\tfrac12h^2\|\partial_h\widehat f_h\|_2^2.$$

The Q-hat criterion has no lambda parameter, so there is one Q-hat distribution per sample size. Its minimizers are computed continuously, rather than selected from the 201 candidate bandwidths. Only the histogram uses the same geometric-grid bins as the spectral distributions.

**Run setup, then the three cells below.** The first loads four completed thesis runs ($M=1000$ each). The second regenerates the identical seeded Gaussian sample bank and computes or resumes the continuous Q-hat minimizers. It does not recompute spectral selections. The third plots the five distributions. `OVERLAY_RUN_IDS` identifies the archived runs independently of the current `ADDITIONAL_LAMBDA_VALUES` setting.

The analytic Gaussian-pair formula follows the existing [joint-limit notebook's `local_beta2_log_metric`](../xx_notebooks_asymptotics/experiment_1_joint_N_lambda_limit.ipynb), also described at [the thesis's joint-limit discussion and `prop:main_thm`](../../../Abschlussarbeit_2021/LaTeX/chapters/5_bandwidth_selection.tex). [qhat.py](../kde_tools/qhat.py) differentiates that formula analytically in log h. A finely binned Fourier scan supplies starting brackets only; all final candidate minima and both endpoints are evaluated using the analytic Gaussian-pair formula. This is a numerical search over a finite interval. The final saved validation repeats it with doubled scan and Fourier resolution on three repetitions at every sample size.

Sources for the paired samples and spectral frequencies: the saved `settings.json` and `frequencies.csv` in the [lambda 1 run](../outputs/data/fast_l2_bandwidth_distributions_vs_n_lambda/thesis/028fa8664f9c), [lambda 10 run](../outputs/data/fast_l2_bandwidth_distributions_vs_n_lambda/thesis/1de0f2734d87), [lambda 100 run](../outputs/data/fast_l2_bandwidth_distributions_vs_n_lambda/thesis/4e41e6f126cf), and [lambda 1000 run](../outputs/data/fast_l2_bandwidth_distributions_vs_n_lambda/thesis/d7062cba5620).

Continuous minima, histogram frequencies, summaries, settings, and refinement diagnostics are saved separately under `outputs/data/fast_l2_bandwidth_distributions_vs_n_lambda/thesis/continuous_qhat/`. The new PNG includes `with_continuous_qhat` in its name; the automatic PDF has a separate descriptive filename, preserving the earlier spectral-only plot.


```python
# Load the completed runs, without calculating KDEs or eigenvectors again.
OVERLAY_PROFILE = "thesis"
OVERLAY_RUN_IDS = {
    1.0: "028fa8664f9c",
    10.0: "1de0f2734d87",
    100.0: "4e41e6f126cf",
    1000.0: "d7062cba5620",
}
overlay_sample_sizes = list(range(100, 1001, 100))
overlay_frequencies = {}
overlay_common_configuration = None

for overlay_lambda, overlay_run_id in OVERLAY_RUN_IDS.items():
    overlay_source_directory = data_output_directory / OVERLAY_PROFILE / overlay_run_id
    overlay_manifest = json.loads((overlay_source_directory / "settings.json").read_text())
    if (overlay_manifest["profile"] != OVERLAY_PROFILE
            or overlay_manifest["lambda_values"] != [overlay_lambda]
            or overlay_manifest["sample_sizes"] != overlay_sample_sizes):
        raise ValueError(f"Unexpected experiment configuration: {overlay_source_directory}")
    overlay_configuration = {
        key: value for key, value in overlay_manifest.items() if key != "lambda_values"
    }
    if overlay_common_configuration is None:
        overlay_common_configuration = overlay_configuration
        overlay_settings = dict(overlay_manifest["settings"])
        overlay_h_grid = bandwidth_grid_linear_in_psi_beta(
            h_min=overlay_settings["h_min"], h_max=overlay_settings["h_max"],
            beta=2.0, grid_size=overlay_settings["grid_size"],
        )
        overlay_repetitions = int(overlay_settings["repetitions"])
    elif overlay_configuration != overlay_common_configuration:
        raise ValueError("The four overlay runs must share all settings except lambda.")

    with (overlay_source_directory / "frequencies.csv").open(newline="") as overlay_stream:
        overlay_rows = list(csv.DictReader(overlay_stream))
    if len(overlay_rows) != len(overlay_sample_sizes) * len(overlay_h_grid):
        raise ValueError(f"Incomplete frequency table: {overlay_source_directory}")
    if any(float(row["lambda"]) != overlay_lambda for row in overlay_rows):
        raise ValueError(f"Unexpected lambda in frequency table: {overlay_source_directory}")
    overlay_frequencies[overlay_lambda] = {}
    for overlay_n in overlay_sample_sizes:
        overlay_point_rows = sorted(
            (row for row in overlay_rows if int(row["n"]) == overlay_n),
            key=lambda row: int(row["grid_index"]),
        )
        if [int(row["grid_index"]) for row in overlay_point_rows] != list(range(len(overlay_h_grid))):
            raise ValueError(f"Missing or duplicate grid positions at n={overlay_n}.")
        if not np.allclose(
            [float(row["h"]) for row in overlay_point_rows], overlay_h_grid,
            rtol=0, atol=1e-15,
        ):
            raise ValueError("The saved bandwidth grid differs from the overlay grid.")
        overlay_counts = np.array([int(row["count"]) for row in overlay_point_rows])
        if np.any(overlay_counts < 0) or overlay_counts.sum() != overlay_repetitions:
            raise ValueError(f"Incomplete distribution: lambda={overlay_lambda:g}, n={overlay_n}.")
        overlay_frequency = overlay_counts / overlay_repetitions
        if not np.allclose(
            overlay_frequency, [float(row["relative_frequency"]) for row in overlay_point_rows],
            rtol=0, atol=1e-15,
        ):
            raise ValueError("Saved frequencies do not agree with the repetition counts.")
        overlay_frequencies[overlay_lambda][overlay_n] = overlay_frequency
    print(f"Loaded lambda={overlay_lambda:g}: {overlay_source_directory}")

print(f"Ready: ten panels, four lambdas per panel, M={overlay_repetitions}, N={len(overlay_h_grid)}.")
```

```python
from concurrent.futures import ThreadPoolExecutor
from threadpoolctl import threadpool_limits
from kde_tools.qhat import QHAT_VERSION, continuous_qhat_minimum
from kde_tools.continuous_runs import CONTINUOUS_RUNNER_VERSION, run_continuous_point

QHAT_WORKERS = 4
QHAT_SCAN_POINTS = 257
QHAT_SPATIAL_POINTS = 32768
QHAT_HALF_WIDTH = 16.0
QHAT_REUSE_CHECKPOINTS = True

overlay_qhat_samples = np.random.default_rng(overlay_settings["seed"]).normal(
    0.0, np.sqrt(overlay_settings["sigma2"]),
    size=(overlay_repetitions, overlay_settings["maximum_sample_size"]),
)
qhat_configuration = {
    "version": QHAT_VERSION, "runner": CONTINUOUS_RUNNER_VERSION,
    "source_sha256": hashlib.sha256((code_directory / "kde_tools/qhat.py").read_bytes()).hexdigest(),
    "sampling_configuration": overlay_common_configuration,
    "sample_bank_sha256": hashlib.sha256(overlay_qhat_samples.tobytes()).hexdigest(),
    "h_min": float(overlay_h_grid[0]), "h_max": float(overlay_h_grid[-1]),
    "scan_points": QHAT_SCAN_POINTS, "spatial_points": QHAT_SPATIAL_POINTS,
    "half_width": QHAT_HALF_WIDTH,
}
qhat_configuration_json = json.dumps(qhat_configuration, sort_keys=True)
qhat_configuration_id = hashlib.sha256(qhat_configuration_json.encode()).hexdigest()[:12]
qhat_data_directory = (data_output_directory / OVERLAY_PROFILE / "continuous_qhat"
                       / qhat_configuration_id)
qhat_data_directory.mkdir(parents=True, exist_ok=True)
(qhat_data_directory / "settings.json").write_text(json.dumps(qhat_configuration, indent=2) + "\n")


continuous_cache_root = code_directory / "outputs/data/shared_continuous_minimizers"
continuous_import_directories = [
    *sorted((data_output_directory / OVERLAY_PROFILE / "continuous_qhat").glob("*")),
    *sorted((code_directory / "outputs/data/continuous_ise_qhat_distributions_vs_n/thesis").glob("*")),
]
qhat_search_settings = {key: qhat_configuration[key] for key in
                        ("h_min", "h_max", "scan_points", "spatial_points", "half_width")}
qhat_search_settings["sigma2"] = overlay_settings["sigma2"]


def run_qhat_point(current_n, *, reuse=True):
    shared = run_continuous_point(
        current_n, overlay_qhat_samples, qhat_search_settings, ("Q0",), qhat_data_directory,
        cache_root=continuous_cache_root, configuration=qhat_configuration,
        workers=QHAT_WORKERS, reuse=reuse, import_directories=continuous_import_directories,
    )
    result = {field: shared["values"][:, 0, column] for column, field in enumerate(shared["fields"])}
    result["qhat"] = result.pop("objective")
    result["boundary"] = (result["boundary"] != 0).astype(float)
    return {field: result[field] for field in
            ("h", "qhat", "log_derivative", "log_curvature", "boundary", "interior_candidates")}


overlay_qhat_results = {}
with threadpool_limits(limits=1):
    for qhat_n in overlay_sample_sizes:
        overlay_qhat_results[qhat_n] = run_qhat_point(qhat_n, reuse=QHAT_REUSE_CHECKPOINTS)

overlay_qhat_frequencies = {}
qhat_selection_rows, qhat_frequency_rows, qhat_summary_rows = [], [], []
qhat_log_positions = np.arange(len(overlay_h_grid), dtype=float)
qhat_bin_edges = np.arange(len(overlay_h_grid) + 1) - 0.5
for qhat_n in overlay_sample_sizes:
    result = overlay_qhat_results[qhat_n]
    positions = np.interp(np.log(result["h"]), np.log(overlay_h_grid), qhat_log_positions)
    counts, _ = np.histogram(positions, bins=qhat_bin_edges)
    assert counts.sum() == overlay_repetitions
    overlay_qhat_frequencies[qhat_n] = counts / overlay_repetitions
    for repetition in range(overlay_repetitions):
        qhat_selection_rows.append({"n": qhat_n, "repetition": repetition,
                                    **{field: float(values[repetition]) for field, values in result.items()}})
    for index in range(len(overlay_h_grid)):
        qhat_frequency_rows.append({"n": qhat_n, "grid_index": index,
                                   "h": float(overlay_h_grid[index]), "count": int(counts[index]),
                                   "relative_frequency": float(counts[index] / overlay_repetitions)})
    qhat_summary_rows.append({
        "n": qhat_n, "M": overlay_repetitions, "mean_h": float(np.mean(result["h"])),
        "median_h": float(np.median(result["h"])), "sd_h": float(np.std(result["h"], ddof=1)),
        "q05_h": float(np.quantile(result["h"], .05)), "q95_h": float(np.quantile(result["h"], .95)),
        "boundary_count": int(np.sum(result["boundary"])),
        "max_interior_candidates": int(np.max(result["interior_candidates"])),
    })
for filename, rows in (("continuous_minimizers.csv", qhat_selection_rows),
                       ("frequencies.csv", qhat_frequency_rows), ("summary.csv", qhat_summary_rows)):
    with (qhat_data_directory / filename).open("w", newline="") as stream:
        writer = csv.DictWriter(stream, fieldnames=rows[0].keys())
        writer.writeheader()
        writer.writerows(rows)

# Check the numerical search with twice as many scan nodes and Fourier nodes.
qhat_refinement_checks = []
with threadpool_limits(limits=1):
    for qhat_n in overlay_sample_sizes:
        for repetition in sorted({0, overlay_repetitions // 2, overlay_repetitions - 1}):
            refined = continuous_qhat_minimum(
                overlay_qhat_samples[repetition, :qhat_n],
                h_min=qhat_configuration["h_min"], h_max=qhat_configuration["h_max"],
                scan_points=2 * QHAT_SCAN_POINTS - 1, spatial_points=2 * QHAT_SPATIAL_POINTS,
                half_width=QHAT_HALF_WIDTH,
            )
            difference = abs(refined["h"] - overlay_qhat_results[qhat_n]["h"][repetition])
            if difference > 1e-7:
                raise ArithmeticError("Continuous Q-hat minimizer changed under scan refinement.")
            qhat_refinement_checks.append({"n": qhat_n, "repetition": repetition,
                                           "bandwidth_discrepancy": float(difference)})
qhat_validation = {
    "completed_minimizers": len(qhat_selection_rows),
    "refinement_checks": qhat_refinement_checks,
    "maximum_refinement_discrepancy": max(row["bandwidth_discrepancy"] for row in qhat_refinement_checks),
    "summary": qhat_summary_rows,
}
(qhat_data_directory / "validation.json").write_text(json.dumps(qhat_validation, indent=2) + "\n")
print(f"Continuous Q-hat minima and diagnostics saved: {qhat_data_directory}")
```
