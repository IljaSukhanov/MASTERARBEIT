# Why Raw Distance Matrices Are Bad Perron-Vector Inputs

This notebook tests the phenomenon behind the lemma from the notes: when the Perron vector is computed from a raw distance matrix, the maximal entry tends to occur at an extreme point of the ordered scale family. That is undesirable for scale selection, because it makes the method prefer boundary scales rather than an intrinsic/interior scale.

The notebook compares two choices:

1. raw distance matrix

$$
D_{ij}=d(x_i,x_j),
$$

2. transformed similarity matrix

$$
S_{ij}=\exp(-\gamma D_{ij}).
$$

The distance matrix rewards being far away from many other scales. The similarity matrix rewards being close to many other scales. This reversal is the main reason the Perron vector of a distance matrix is usually not the right object for selecting a representative bandwidth.

```python
from __future__ import annotations

import os
from pathlib import Path

os.environ.setdefault("MPLCONFIGDIR", str(Path.cwd() / ".matplotlib-cache"))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from kde_tools.bandwidths import bandwidth_grid_linear_in_psi_beta

plt.style.use("seaborn-v0_8-whitegrid")
np.set_printoptions(precision=4, suppress=True)
```

## Shared helper functions

```python
def perron_vector_symmetric(matrix: np.ndarray) -> tuple[float, np.ndarray]:
    """Largest eigenvalue and sign-normalized Perron vector of a symmetric matrix."""
    matrix = np.asarray(matrix, dtype=float)
    eigenvalues, eigenvectors = np.linalg.eigh((matrix + matrix.T) / 2.0)
    index = int(np.argmax(eigenvalues))
    vector = eigenvectors[:, index]
    if vector.sum() < 0:
        vector = -vector
    if np.any(vector < -1e-10):
        vector = np.abs(vector)
    if vector.sum() > 0:
        vector = vector / vector.sum()
    return float(eigenvalues[index]), vector


def exp_similarity(distance_matrix: np.ndarray, gamma: float = 1.0) -> tuple[np.ndarray, float]:
    gamma_value = float(gamma)
    if not np.isfinite(gamma_value) or gamma_value <= 0.0:
        raise ValueError("gamma must be finite and strictly positive.")
    return np.exp(-gamma_value * distance_matrix), gamma_value


def endpoint_label(index: int, n_points: int) -> str:
    if index == 0:
        return "left boundary"
    if index == n_points - 1:
        return "right boundary"
    return "interior"


def summarize_selection(name: str, grid: np.ndarray, vector: np.ndarray) -> dict[str, object]:
    index = int(np.argmax(vector))
    return {
        "matrix": name,
        "selected_index": index,
        "selected_value": float(grid[index]),
        "location": endpoint_label(index, len(grid)),
        "max_weight": float(vector[index]),
    }


def plot_matrix_and_perron(matrix: np.ndarray, grid: np.ndarray, title: str, ax_matrix, ax_vector):
    _, vector = perron_vector_symmetric(matrix)
    selected_index = int(np.argmax(vector))
    image = ax_matrix.imshow(matrix, origin="lower", aspect="auto", cmap="viridis")
    ax_matrix.set_title(title)
    ax_matrix.set_xlabel("index j")
    ax_matrix.set_ylabel("index i")
    plt.colorbar(image, ax=ax_matrix, fraction=0.046, pad=0.04)

    ax_vector.plot(grid, vector, marker="o", markersize=3, lw=2)
    ax_vector.axvline(grid[selected_index], color="tab:red", ls="--", lw=2)
    ax_vector.set_title(f"Perron vector, selected {endpoint_label(selected_index, len(grid))}")
    ax_vector.set_xlabel("ordered scale value")
    ax_vector.set_ylabel("normalized weight")
    return vector
```

## Experiment 1: ordered points on an interval

Take points on a line segment,

$$
x_1<\dots<x_N\in[0,1],
$$

and define

$$
D_{ij}=|x_i-x_j|.
$$

The distance-matrix Perron vector should prefer one of the endpoints. This is the simplest finite-dimensional version of the boundary-selection problem.

```python
n_grid_interval = 41
interval_grid = np.linspace(0.0, 1.0, n_grid_interval)
interval_distance = np.abs(interval_grid[:, None] - interval_grid[None, :])
interval_similarity, interval_gamma = exp_similarity(interval_distance, gamma=1.0)

_, interval_distance_vector = perron_vector_symmetric(interval_distance)
_, interval_similarity_vector = perron_vector_symmetric(interval_similarity)

summary_interval = pd.DataFrame([
    summarize_selection("raw distance D", interval_grid, interval_distance_vector),
    summarize_selection(r"similarity exp(-gamma D)", interval_grid, interval_similarity_vector),
])

fig, axes = plt.subplots(2, 2, figsize=(13, 8))
plot_matrix_and_perron(interval_distance, interval_grid, "Raw distance matrix", axes[0, 0], axes[0, 1])
plot_matrix_and_perron(interval_similarity, interval_grid, rf"Similarity matrix, gamma={interval_gamma:.3g}", axes[1, 0], axes[1, 1])
fig.tight_layout()

summary_interval
```

## Experiment 2: power distances

The endpoint effect is not special to the linear distance. We also test

$$
D^{(p)}_{ij}=|x_i-x_j|^p.
$$

For several exponents, the raw distance matrix still selects a boundary point.

```python
power_results = []
fig, axes = plt.subplots(1, 4, figsize=(16, 3.5), sharey=True)

for ax, power in zip(axes, [0.5, 1.0, 1.5, 2.0]):
    D_power = interval_distance ** power
    _, vector_power = perron_vector_symmetric(D_power)
    selected_index = int(np.argmax(vector_power))
    ax.plot(interval_grid, vector_power, marker="o", markersize=3, lw=2)
    ax.axvline(interval_grid[selected_index], color="tab:red", ls="--", lw=2)
    ax.set_title(rf"$|x-y|^{{{power:g}}}$")
    ax.set_xlabel("x")
    power_results.append(summarize_selection(rf"|x-y|^{power:g}", interval_grid, vector_power))

axes[0].set_ylabel("normalized Perron weight")
fig.tight_layout()

pd.DataFrame(power_results)
```

## Exact Gaussian KDE distance matrices

Now we repeat the test for bandwidths. For one sample from

$$
X_1,\dots,X_n\sim N(0,\sigma^2),
$$

we compute Gaussian KDEs \(\hat f_h\) on a bandwidth grid. The squared \(L^2\) distance is computed exactly by Gaussian convolution:

$$
\langle \hat f_{h_i},\hat f_{h_j}\rangle
=
\frac{1}{n^2}\sum_{k,l=1}^n
\varphi_{\sqrt{h_i^2+h_j^2}}(X_k-X_l).
$$

Then

$$
\|\hat f_{h_i}-\hat f_{h_j}\|_2^2=I_{ii}+I_{jj}-2I_{ij}.
$$

We compare Perron vectors of

- raw squared distance \(D_{ij}=\|\hat f_{h_i}-\hat f_{h_j}\|_2^2\),
- raw distance \(D_{ij}=\|\hat f_{h_i}-\hat f_{h_j}\|_2\),
- similarity \(S_{ij}=\exp(-\gamma\|\hat f_{h_i}-\hat f_{h_j}\|_2^2)\),
- oracle ISE benchmark.

```python
def gaussian_pdf_sd(x: np.ndarray, sd: float) -> np.ndarray:
    return np.exp(-0.5 * (x / sd) ** 2) / (sd * np.sqrt(2.0 * np.pi))


def bandwidth_grid_for_gaussian_experiment(n: int, sigma2: float, n_h: int = 60) -> np.ndarray:
    sigma = np.sqrt(sigma2)
    h_min = sigma * 0.25 * n ** (-0.45)
    h_max = sigma * 3.00 * n ** (-0.08)
    return bandwidth_grid_linear_in_psi_beta(
        h_min=h_min,
        h_max=h_max,
        beta=2.0,
        grid_size=n_h,
    )


def exact_gaussian_kde_l2_gram(samples: np.ndarray, h_grid: np.ndarray) -> np.ndarray:
    samples = np.asarray(samples, dtype=float)
    h_grid = np.asarray(h_grid, dtype=float)
    pairwise_differences = samples[:, None] - samples[None, :]
    n_h = len(h_grid)
    gram = np.empty((n_h, n_h), dtype=float)

    for i, h_i in enumerate(h_grid):
        for j in range(i, n_h):
            h_j = h_grid[j]
            sd = np.sqrt(h_i**2 + h_j**2)
            value = gaussian_pdf_sd(pairwise_differences, sd).mean()
            gram[i, j] = value
            gram[j, i] = value
    return gram


def squared_l2_distance_from_gram(gram: np.ndarray) -> np.ndarray:
    diagonal = np.diag(gram)
    return np.maximum(diagonal[:, None] + diagonal[None, :] - 2.0 * gram, 0.0)


def exact_kde_true_normal_inner(samples: np.ndarray, h_grid: np.ndarray, sigma2: float) -> np.ndarray:
    samples = np.asarray(samples, dtype=float)
    return np.array([
        gaussian_pdf_sd(samples, np.sqrt(h**2 + sigma2)).mean()
        for h in h_grid
    ])


def exact_oracle_ise_values(samples: np.ndarray, h_grid: np.ndarray, sigma2: float, gram: np.ndarray | None = None) -> np.ndarray:
    if gram is None:
        gram = exact_gaussian_kde_l2_gram(samples, h_grid)
    kde_norm_sq = np.diag(gram)
    kde_true_inner = exact_kde_true_normal_inner(samples, h_grid, sigma2=sigma2)
    sigma = np.sqrt(sigma2)
    true_norm_sq = 1.0 / (2.0 * np.sqrt(np.pi) * sigma)
    return kde_norm_sq - 2.0 * kde_true_inner + true_norm_sq
```

```python
rng = np.random.default_rng(20260715)
n_kde = 200
sigma2_kde = 1.0
n_h_kde = 70

samples = rng.normal(loc=0.0, scale=np.sqrt(sigma2_kde), size=n_kde)
h_grid = bandwidth_grid_for_gaussian_experiment(n=n_kde, sigma2=sigma2_kde, n_h=n_h_kde)
gram = exact_gaussian_kde_l2_gram(samples, h_grid)
D_l2_squared = squared_l2_distance_from_gram(gram)
D_l2 = np.sqrt(D_l2_squared)
S_gamma, gamma_kde = exp_similarity(D_l2_squared, gamma=1.0)
ise_values = exact_oracle_ise_values(samples, h_grid, sigma2=sigma2_kde, gram=gram)
oracle_index = int(np.argmin(ise_values))

_, v_D_squared = perron_vector_symmetric(D_l2_squared)
_, v_D = perron_vector_symmetric(D_l2)
_, v_S = perron_vector_symmetric(S_gamma)

kde_summary = pd.DataFrame([
    summarize_selection(r"raw squared L2 distance", h_grid, v_D_squared),
    summarize_selection(r"raw L2 distance", h_grid, v_D),
    summarize_selection(r"similarity exp(-gamma L2^2)", h_grid, v_S),
    {
        "matrix": "oracle ISE minimum",
        "selected_index": oracle_index,
        "selected_value": float(h_grid[oracle_index]),
        "location": endpoint_label(oracle_index, len(h_grid)),
        "max_weight": np.nan,
    },
])

fig, axes = plt.subplots(2, 3, figsize=(17, 8))
plot_matrix_and_perron(D_l2_squared, h_grid, r"Raw squared $L^2$ distance", axes[0, 0], axes[1, 0])
plot_matrix_and_perron(D_l2, h_grid, r"Raw $L^2$ distance", axes[0, 1], axes[1, 1])
plot_matrix_and_perron(S_gamma, h_grid, rf"Similarity $e^{{-\gamma D}}$, gamma={gamma_kde:.3g}", axes[0, 2], axes[1, 2])
for ax in axes[1]:
    ax.set_xscale("log")
    ax.axvline(h_grid[oracle_index], color="tab:green", ls=":", lw=2, label="oracle")
    ax.legend()
fig.tight_layout()

kde_summary
```

## Repeated KDE diagnostic

This block repeats the KDE test. The important output is the boundary-selection rate. A raw distance matrix is considered bad if it frequently selects index 0 or index \(N-1\).

```python
n_repetitions = 20
n_repeated = 100
sigma2_repeated = 1.0
n_h_repeated = 45
base_seed = 910_000

h_grid_repeated = bandwidth_grid_for_gaussian_experiment(n=n_repeated, sigma2=sigma2_repeated, n_h=n_h_repeated)
rows = []

for repetition in range(n_repetitions):
    rng = np.random.default_rng(base_seed + repetition)
    x = rng.normal(loc=0.0, scale=np.sqrt(sigma2_repeated), size=n_repeated)
    gram_current = exact_gaussian_kde_l2_gram(x, h_grid_repeated)
    D2_current = squared_l2_distance_from_gram(gram_current)
    D_current = np.sqrt(D2_current)
    S_current, gamma_current = exp_similarity(D2_current, gamma=1.0)
    ise_current = exact_oracle_ise_values(x, h_grid_repeated, sigma2=sigma2_repeated, gram=gram_current)

    selectors = {
        "raw squared L2 distance": int(np.argmax(perron_vector_symmetric(D2_current)[1])),
        "raw L2 distance": int(np.argmax(perron_vector_symmetric(D_current)[1])),
        "similarity exp(-gamma L2^2)": int(np.argmax(perron_vector_symmetric(S_current)[1])),
        "oracle ISE": int(np.argmin(ise_current)),
    }
    for name, index in selectors.items():
        rows.append({
            "repetition": repetition,
            "selector": name,
            "selected_index": index,
            "selected_h": float(h_grid_repeated[index]),
            "is_boundary": index in {0, len(h_grid_repeated) - 1},
            "gamma": gamma_current,
        })

repeated_results = pd.DataFrame(rows)
summary_repeated = (
    repeated_results.groupby("selector")
    .agg(
        boundary_rate=("is_boundary", "mean"),
        median_index=("selected_index", "median"),
        median_h=("selected_h", "median"),
        mean_h=("selected_h", "mean"),
    )
    .reset_index()
)

fig, ax = plt.subplots(figsize=(11, 5))
for selector, group in repeated_results.groupby("selector"):
    frequencies = np.bincount(group["selected_index"].to_numpy(dtype=int), minlength=len(h_grid_repeated)) / n_repetitions
    ax.plot(h_grid_repeated, frequencies, marker="o", markersize=4, lw=2, label=selector)
ax.set_xscale("log")
ax.set_title(f"Selected bandwidth distributions over {n_repetitions} repetitions")
ax.set_xlabel("bandwidth h")
ax.set_ylabel("relative frequency")
ax.legend()
fig.tight_layout()

summary_repeated
```

## Interpretation

The raw distance matrix answers the wrong question. A point has a large Perron weight if it is far from many other points. In an ordered family of scales, the points farthest from many other scales are typically the endpoints.

For bandwidth selection, this is exactly the opposite of what we want: we do not want the most extreme bandwidth. We want a representative or structurally central bandwidth. A similarity matrix such as

$$
S_{ij}=\exp(-\gamma D_{ij})
$$

reverses the logic: a point receives a large Perron weight if it is similar to many other points. This is why similarity matrices are more appropriate for Perron-vector based scale selection.

```python
from __future__ import annotations

import numpy as np
from numpy.typing import ArrayLike
from scipy.special import logsumexp


def log_gaussian_kde(
    evaluation_points: ArrayLike,
    data: ArrayLike,
    bandwidth: float,
    batch_size: int = 10_000,
) -> np.ndarray:
    """
    Evaluate the log-density of a one-dimensional Gaussian KDE.

    The KDE is

        f_h(x) = (1 / n) * sum_i N(x | X_i, h^2).

    Parameters
    ----------
    evaluation_points:
        Points at which the KDE is evaluated.
    data:
        One-dimensional sample used to construct the KDE.
    bandwidth:
        Positive Gaussian bandwidth h.
    batch_size:
        Number of evaluation points processed simultaneously.

    Returns
    -------
    np.ndarray
        Log-KDE values at the evaluation points.
    """
    x = np.asarray(evaluation_points, dtype=float).reshape(-1)
    sample = np.asarray(data, dtype=float).reshape(-1)

    if sample.size == 0:
        raise ValueError("data must contain at least one observation.")
    if bandwidth <= 0:
        raise ValueError("bandwidth must be strictly positive.")
    if batch_size <= 0:
        raise ValueError("batch_size must be strictly positive.")

    n = sample.size
    normalizing_constant = (
        -np.log(n)
        - np.log(bandwidth)
        - 0.5 * np.log(2.0 * np.pi)
    )

    result = np.empty(x.size, dtype=float)

    for start in range(0, x.size, batch_size):
        stop = min(start + batch_size, x.size)

        # Shape: (number of evaluation points in batch, sample size)
        standardized_differences = (
            x[start:stop, None] - sample[None, :]
        ) / bandwidth

        log_kernel_values = -0.5 * standardized_differences**2

        result[start:stop] = (
            normalizing_constant
            + logsumexp(log_kernel_values, axis=1)
        )

    return result


def kl_between_gaussian_kdes_mc(
    data: ArrayLike,
    bandwidth_p: float,
    bandwidth_q: float,
    n_simulations: int = 100_000,
    random_state: int | np.random.Generator | None = None,
    batch_size: int = 10_000,
) -> dict[str, float]:
    """
    Estimate KL(f_{h_p} || f_{h_q}) by Monte Carlo simulation.

    Both KDEs are constructed from the same data but use different
    Gaussian bandwidths.

    A draw from f_{h_p} is generated by:
        1. selecting one observation X_I uniformly,
        2. drawing Z ~ N(0, 1),
        3. setting Y = X_I + h_p Z.

    Then

        KL(f_{h_p} || f_{h_q})
        = E_{Y ~ f_{h_p}}[
              log f_{h_p}(Y) - log f_{h_q}(Y)
          ].

    Parameters
    ----------
    data:
        One-dimensional sample.
    bandwidth_p:
        Bandwidth of the KDE in the first KL argument.
    bandwidth_q:
        Bandwidth of the KDE in the second KL argument.
    n_simulations:
        Number of Monte Carlo samples.
    random_state:
        Integer seed, NumPy Generator, or None.
    batch_size:
        Batch size used when evaluating log-KDEs.

    Returns
    -------
    dict
        Dictionary containing the KL estimate, Monte Carlo standard error,
        and an approximate 95% confidence interval.
    """
    sample = np.asarray(data, dtype=float).reshape(-1)

    if sample.size == 0:
        raise ValueError("data must contain at least one observation.")
    if bandwidth_p <= 0 or bandwidth_q <= 0:
        raise ValueError("Both bandwidths must be strictly positive.")
    if n_simulations <= 1:
        raise ValueError("n_simulations must be greater than 1.")

    if isinstance(random_state, np.random.Generator):
        rng = random_state
    else:
        rng = np.random.default_rng(random_state)

    component_indices = rng.integers(
        low=0,
        high=sample.size,
        size=n_simulations,
    )

    simulated_points = (
        sample[component_indices]
        + bandwidth_p * rng.standard_normal(n_simulations)
    )

    log_p = log_gaussian_kde(
        evaluation_points=simulated_points,
        data=sample,
        bandwidth=bandwidth_p,
        batch_size=batch_size,
    )

    log_q = log_gaussian_kde(
        evaluation_points=simulated_points,
        data=sample,
        bandwidth=bandwidth_q,
        batch_size=batch_size,
    )

    log_density_ratio = log_p - log_q

    estimate = float(np.mean(log_density_ratio))
    standard_error = float(
        np.std(log_density_ratio, ddof=1)
        / np.sqrt(n_simulations)
    )

    confidence_radius = 1.96 * standard_error

    return {
        "estimate": estimate,
        "standard_error": standard_error,
        "ci_lower_95": estimate - confidence_radius,
        "ci_upper_95": estimate + confidence_radius,
    }
```

```python
rng = np.random.default_rng(42)
data = rng.normal(loc=0.0, scale=1.0, size=200)

kl_02_05 = kl_between_gaussian_kdes_mc(
    data=data,
    bandwidth_p=0.2,
    bandwidth_q=0.5,
    n_simulations=100_000,
    random_state=123,
)

kl_05_02 = kl_between_gaussian_kdes_mc(
    data=data,
    bandwidth_p=0.5,
    bandwidth_q=0.2,
    n_simulations=100_000,
    random_state=456,
)

print("KL(f_0.2 || f_0.5):", kl_02_05)
print("KL(f_0.5 || f_0.2):", kl_05_02)
```

```python
def kl_kde_matrix_mc(
    data: ArrayLike,
    bandwidths: ArrayLike,
    n_simulations: int = 50_000,
    random_state: int | None = None,
    batch_size: int = 10_000,
) -> tuple[np.ndarray, np.ndarray]:
    """
    Estimate the directed KL matrix for Gaussian KDEs.

    Returns
    -------
    kl_matrix:
        Entry (i, j) estimates KL(f_{h_i} || f_{h_j}).
    standard_errors:
        Monte Carlo standard errors for the corresponding entries.
    """
    bandwidths = np.asarray(bandwidths, dtype=float).reshape(-1)

    if np.any(bandwidths <= 0):
        raise ValueError("All bandwidths must be strictly positive.")

    rng = np.random.default_rng(random_state)
    m = bandwidths.size

    kl_matrix = np.zeros((m, m), dtype=float)
    standard_errors = np.zeros((m, m), dtype=float)

    for i, h_p in enumerate(bandwidths):
        # Reuse one Monte Carlo sample from the i-th KDE for all j.
        sample = np.asarray(data, dtype=float).reshape(-1)
        component_indices = rng.integers(
            0, sample.size, size=n_simulations
        )
        simulated_points = (
            sample[component_indices]
            + h_p * rng.standard_normal(n_simulations)
        )

        log_p = log_gaussian_kde(
            simulated_points,
            sample,
            h_p,
            batch_size=batch_size,
        )

        for j, h_q in enumerate(bandwidths):
            if i == j:
                continue

            log_q = log_gaussian_kde(
                simulated_points,
                sample,
                h_q,
                batch_size=batch_size,
            )

            log_ratio = log_p - log_q

            kl_matrix[i, j] = np.mean(log_ratio)
            standard_errors[i, j] = (
                np.std(log_ratio, ddof=1)
                / np.sqrt(n_simulations)
            )

    return kl_matrix, standard_errors
```

```python
from __future__ import annotations

import numpy as np
import matplotlib.pyplot as plt
from numpy.typing import ArrayLike
from scipy.special import logsumexp

from kde_tools.bandwidths import bandwidth_grid_linear_in_psi_beta


def log_gaussian_kdes(
    evaluation_points: ArrayLike,
    data: ArrayLike,
    bandwidths: ArrayLike,
    point_batch_size: int = 256,
    bandwidth_batch_size: int = 10,
) -> np.ndarray:
    """
    Evaluate several one-dimensional Gaussian KDEs at the same points.

    For bandwidth h_j,

        f_hat_hj(x)
        =
        (1 / n) sum_k phi_hj(x - X_k).

    Parameters
    ----------
    evaluation_points:
        Points x at which all KDEs are evaluated.
    data:
        One-dimensional data sample.
    bandwidths:
        Positive KDE bandwidths.
    point_batch_size:
        Number of evaluation points handled simultaneously.
    bandwidth_batch_size:
        Number of bandwidths handled simultaneously.

    Returns
    -------
    log_densities:
        Array of shape (number of evaluation points, number of bandwidths).
        Entry (m, j) is log f_hat_{h_j}(evaluation_points[m]).
    """
    points = np.asarray(evaluation_points, dtype=float).reshape(-1)
    sample = np.asarray(data, dtype=float).reshape(-1)
    hs = np.asarray(bandwidths, dtype=float).reshape(-1)

    if sample.size == 0:
        raise ValueError("data must contain at least one observation.")

    if np.any(hs <= 0):
        raise ValueError("All bandwidths must be strictly positive.")

    if point_batch_size <= 0 or bandwidth_batch_size <= 0:
        raise ValueError("Batch sizes must be strictly positive.")

    number_of_points = points.size
    number_of_bandwidths = hs.size
    sample_size = sample.size

    result = np.empty(
        (number_of_points, number_of_bandwidths),
        dtype=float,
    )

    constants = (
        -np.log(sample_size)
        - np.log(hs)
        - 0.5 * np.log(2.0 * np.pi)
    )

    for point_start in range(0, number_of_points, point_batch_size):
        point_stop = min(
            point_start + point_batch_size,
            number_of_points,
        )

        point_chunk = points[point_start:point_stop]

        # Shape: (point batch size, sample size)
        squared_distances = (
            point_chunk[:, None] - sample[None, :]
        ) ** 2

        for bandwidth_start in range(
            0,
            number_of_bandwidths,
            bandwidth_batch_size,
        ):
            bandwidth_stop = min(
                bandwidth_start + bandwidth_batch_size,
                number_of_bandwidths,
            )

            bandwidth_chunk = hs[bandwidth_start:bandwidth_stop]

            # Shape:
            # (point batch size, bandwidth batch size, sample size)
            log_kernel_values = (
                -squared_distances[:, None, :]
                / (
                    2.0
                    * bandwidth_chunk[None, :, None] ** 2
                )
            )

            result[
                point_start:point_stop,
                bandwidth_start:bandwidth_stop,
            ] = (
                constants[None, bandwidth_start:bandwidth_stop]
                + logsumexp(log_kernel_values, axis=2)
            )

    return result


def gaussian_kde_kl_matrix_mc(
    data: ArrayLike,
    bandwidths: ArrayLike,
    n_mc: int = 1_000,
    random_state: int | None = None,
    point_batch_size: int = 256,
    bandwidth_batch_size: int = 10,
    show_progress: bool = True,
) -> tuple[np.ndarray, np.ndarray]:
    """
    Estimate the directed KL matrix between Gaussian KDEs.

    Entry (i, j) estimates

        KL(f_hat_{h_i} || f_hat_{h_j}).

    For each source KDE f_hat_{h_i}, draws are generated as

        Y = X_I + h_i Z,

    where I is uniform on {1, ..., n} and Z ~ N(0, 1).

    Parameters
    ----------
    data:
        One-dimensional sample.
    bandwidths:
        Positive KDE bandwidths.
    n_mc:
        Number of Monte Carlo draws per source bandwidth.
    random_state:
        Random seed.
    point_batch_size:
        Point batching used during KDE evaluation.
    bandwidth_batch_size:
        Bandwidth batching used during KDE evaluation.
    show_progress:
        Print progress information.

    Returns
    -------
    kl_matrix:
        Directed KL matrix.
    standard_error_matrix:
        Monte Carlo standard error for every entry.
    """
    sample = np.asarray(data, dtype=float).reshape(-1)
    hs = np.asarray(bandwidths, dtype=float).reshape(-1)

    if sample.size == 0:
        raise ValueError("data must contain at least one observation.")

    if np.any(hs <= 0):
        raise ValueError("All bandwidths must be strictly positive.")

    if n_mc <= 1:
        raise ValueError("n_mc must be greater than 1.")

    rng = np.random.default_rng(random_state)

    number_of_bandwidths = hs.size
    sample_size = sample.size

    kl_matrix = np.zeros(
        (number_of_bandwidths, number_of_bandwidths),
        dtype=float,
    )

    standard_error_matrix = np.zeros_like(kl_matrix)

    # Common random numbers across rows.
    # For each h_i, X_I + h_i Z still has exactly the correct KDE law.
    component_indices = rng.integers(
        low=0,
        high=sample_size,
        size=n_mc,
    )
    standard_normals = rng.standard_normal(n_mc)

    for i, source_bandwidth in enumerate(hs):
        simulated_points = (
            sample[component_indices]
            + source_bandwidth * standard_normals
        )

        # log_densities[m, j]
        # = log f_hat_{h_j}(simulated_points[m])
        log_densities = log_gaussian_kdes(
            evaluation_points=simulated_points,
            data=sample,
            bandwidths=hs,
            point_batch_size=point_batch_size,
            bandwidth_batch_size=bandwidth_batch_size,
        )

        source_log_density = log_densities[:, i]

        # Each column j contains
        # log f_hat_{h_i}(Y_m) - log f_hat_{h_j}(Y_m).
        log_ratios = (
            source_log_density[:, None] - log_densities
        )

        kl_matrix[i, :] = np.mean(log_ratios, axis=0)

        standard_error_matrix[i, :] = (
            np.std(log_ratios, axis=0, ddof=1)
            / np.sqrt(n_mc)
        )

        # The diagonal is mathematically exactly zero.
        kl_matrix[i, i] = 0.0
        standard_error_matrix[i, i] = 0.0

        if show_progress and (
            i == 0
            or (i + 1) % 10 == 0
            or i + 1 == number_of_bandwidths
        ):
            print(
                f"Completed row {i + 1}"
                f" of {number_of_bandwidths}"
            )

    return kl_matrix, standard_error_matrix


def population_gaussian_kl_matrix(
    bandwidths: ArrayLike,
    sigma: float,
) -> np.ndarray:
    """
    Population comparison matrix for Gaussian-smoothed N(0, sigma^2).

    If

        f = N(0, sigma^2),

    then

        f * phi_h = N(0, sigma^2 + h^2).

    Entry (i, j) is

        KL(
            N(0, sigma^2 + h_i^2)
            ||
            N(0, sigma^2 + h_j^2)
        ).
    """
    hs = np.asarray(bandwidths, dtype=float).reshape(-1)

    if sigma <= 0:
        raise ValueError("sigma must be strictly positive.")

    variances = sigma**2 + hs**2

    variance_i = variances[:, None]
    variance_j = variances[None, :]

    matrix = 0.5 * (
        variance_i / variance_j
        - 1.0
        + np.log(variance_j / variance_i)
    )

    np.fill_diagonal(matrix, 0.0)
    return matrix


def plot_matrix(
    matrix: np.ndarray,
    bandwidths: np.ndarray,
    title: str,
) -> None:
    """Plot a bandwidth-by-bandwidth matrix."""
    plt.figure(figsize=(8, 7))

    image = plt.imshow(
        matrix,
        origin="lower",
        aspect="auto",
        extent=[
            bandwidths[0],
            bandwidths[-1],
            bandwidths[0],
            bandwidths[-1],
        ],
    )

    plt.colorbar(image, label="KL divergence")
    plt.xlabel(r"Target bandwidth $h_j$")
    plt.ylabel(r"Source bandwidth $h_i$")
    plt.title(title)
    plt.tight_layout()
    plt.show()


# ============================================================
# Experiment
# ============================================================

SEED = 12345

# Data distribution
SIGMA = 1.0
N_DATA = 200

# Bandwidth grid
N_BANDWIDTHS = 100
H_MIN = 0.01
H_MAX = 2.0

# Number of Monte Carlo points for every matrix row.
# Increase this later for greater precision.
N_MC = 1_000

rng = np.random.default_rng(SEED)

# Sample X_1, ..., X_n ~ N(0, sigma^2)
data = rng.normal(
    loc=0.0,
    scale=SIGMA,
    size=N_DATA,
)

# Exactly 100 geometrically spaced interior points of (0.01, 2).
# The endpoints 0.01 and 2 are excluded.
bandwidths = bandwidth_grid_linear_in_psi_beta(
    h_min=H_MIN,
    h_max=H_MAX,
    beta=2.0,
    grid_size=N_BANDWIDTHS + 2,
)[1:-1]

print("First bandwidth:", bandwidths[0])
print("Last bandwidth:", bandwidths[-1])
print("Number of bandwidths:", bandwidths.size)

kl_matrix_mc, kl_standard_errors = gaussian_kde_kl_matrix_mc(
    data=data,
    bandwidths=bandwidths,
    n_mc=N_MC,
    random_state=SEED + 1,
    point_batch_size=256,
    bandwidth_batch_size=10,
    show_progress=True,
)

# Small negative values can occur because of Monte Carlo error.
print(
    "Smallest raw off-diagonal estimate:",
    np.min(
        kl_matrix_mc
        + np.eye(N_BANDWIDTHS) * np.inf
    ),
)

print(
    "Maximum Monte Carlo standard error:",
    np.max(kl_standard_errors),
)

print(
    "Maximum asymmetry |D_ij - D_ji|:",
    np.max(np.abs(kl_matrix_mc - kl_matrix_mc.T)),
)

plot_matrix(
    matrix=kl_matrix_mc,
    bandwidths=bandwidths,
    title=(
        "Monte Carlo KL matrix between Gaussian KDEs\n"
        rf"$n={N_DATA}$, $\sigma={SIGMA}$, "
        rf"$M={N_MC}$"
    ),
)

# Population benchmark
kl_matrix_population = population_gaussian_kl_matrix(
    bandwidths=bandwidths,
    sigma=SIGMA,
)

plot_matrix(
    matrix=kl_matrix_population,
    bandwidths=bandwidths,
    title=(
        "Population KL matrix between Gaussian-smoothed densities\n"
        rf"$N(0,\sigma^2+h_i^2)$, $\sigma={SIGMA}$"
    ),
)

# Difference between the finite-sample KDE matrix and population matrix
plot_matrix(
    matrix=kl_matrix_mc - kl_matrix_population,
    bandwidths=bandwidths,
    title="Finite-sample KDE KL minus population Gaussian KL",
)

# Save results
np.savetxt(
    "outputs/csv/08_kl_distance_matrices/bandwidths.csv",
    bandwidths,
    delimiter=",",
)

np.savetxt(
    "outputs/csv/08_kl_distance_matrices/kl_matrix_mc.csv",
    kl_matrix_mc,
    delimiter=",",
)

np.savetxt(
    "outputs/csv/08_kl_distance_matrices/kl_standard_errors.csv",
    kl_standard_errors,
    delimiter=",",
)

np.savetxt(
    "outputs/csv/08_kl_distance_matrices/kl_matrix_population.csv",
    kl_matrix_population,
    delimiter=",",
)
```

```python
import numpy as np
import matplotlib.pyplot as plt


def positive_perron_vector(
    matrix: np.ndarray,
    tolerance: float = 1e-13,
    max_iterations: int = 100_000,
) -> tuple[float, np.ndarray]:
    """
    Compute the positive right Perron vector by power iteration.

    Returns
    -------
    eigenvalue:
        Perron eigenvalue.
    eigenvector:
        Positive Perron vector normalized to have sum 1.
    """
    matrix = np.asarray(matrix, dtype=float)

    if matrix.ndim != 2 or matrix.shape[0] != matrix.shape[1]:
        raise ValueError("matrix must be square.")

    if np.any(matrix < 0):
        raise ValueError("matrix must be nonnegative.")

    n = matrix.shape[0]
    vector = np.full(n, 1.0 / n)

    for _ in range(max_iterations):
        new_vector = matrix @ vector

        normalization = np.sum(new_vector)

        if normalization <= 0 or not np.isfinite(normalization):
            raise RuntimeError(
                "Power iteration encountered an invalid normalization."
            )

        new_vector /= normalization

        if np.linalg.norm(new_vector - vector, ord=1) < tolerance:
            vector = new_vector
            break

        vector = new_vector
    else:
        raise RuntimeError("Power iteration did not converge.")

    # Rayleigh quotient
    eigenvalue = float(
        vector @ (matrix @ vector)
        / (vector @ vector)
    )

    return eigenvalue, vector


def doob_transform_from_divergence(
    divergence_matrix: np.ndarray,
    tau: float | None = None,
) -> dict[str, np.ndarray | float]:
    """
    Transform a directed divergence matrix into a similarity matrix,
    compute the left/right Perron vectors, the Doob transform, and its
    stationary distribution.

    Parameters
    ----------
    divergence_matrix:
        Matrix D with D[i, j] representing the directed divergence
        from state i to state j.
    tau:
        Similarity scale in

            S[i, j] = exp(-D[i, j] / tau).

        If None, the median positive off-diagonal divergence is used.

    Returns
    -------
    Dictionary containing:
        divergence_matrix
        similarity_matrix
        tau
        perron_eigenvalue
        right_perron_vector
        left_perron_vector
        transition_matrix
        stationary_distribution
    """
    D = np.asarray(divergence_matrix, dtype=float).copy()

    if D.ndim != 2 or D.shape[0] != D.shape[1]:
        raise ValueError("divergence_matrix must be square.")

    if not np.all(np.isfinite(D)):
        raise ValueError("The divergence matrix contains non-finite values.")

    # Monte Carlo estimates can occasionally be slightly negative.
    D = np.maximum(D, 0.0)
    np.fill_diagonal(D, 0.0)

    n = D.shape[0]
    off_diagonal_mask = ~np.eye(n, dtype=bool)
    positive_off_diagonal = D[
        off_diagonal_mask & (D > 0)
    ]

    if tau is None:
        if positive_off_diagonal.size == 0:
            raise ValueError(
                "Cannot determine tau because all divergences are zero."
            )

        tau = float(np.median(positive_off_diagonal))

    if tau <= 0 or not np.isfinite(tau):
        raise ValueError("tau must be finite and strictly positive.")

    # Directed positive similarity matrix
    S = np.exp(-D / tau)

    # Right Perron vector: S r = rho r
    rho_right, right = positive_perron_vector(S)

    # Left Perron vector: S^T ell = rho ell
    rho_left, left = positive_perron_vector(S.T)

    # Use a bilinear estimate of the common Perron eigenvalue.
    rho = float(
        left @ (S @ right)
        / (left @ right)
    )

    # Doob transform:
    #
    # P_ij = S_ij r_j / (rho r_i)
    P = (
        S
        * right[None, :]
        / (rho * right[:, None])
    )

    # Remove negligible numerical row-sum error.
    P /= P.sum(axis=1, keepdims=True)

    # Natural left-right combination
    stationary = left * right
    stationary /= stationary.sum()

    # Diagnostics
    row_sum_error = np.max(
        np.abs(P.sum(axis=1) - 1.0)
    )

    stationarity_error = np.max(
        np.abs(stationary @ P - stationary)
    )

    eigenvalue_difference = abs(rho_right - rho_left)

    print(f"Similarity scale tau: {tau:.8g}")
    print(f"Perron eigenvalue: {rho:.8g}")
    print(
        "Left/right eigenvalue difference:",
        f"{eigenvalue_difference:.3e}",
    )
    print(
        "Maximum transition row-sum error:",
        f"{row_sum_error:.3e}",
    )
    print(
        "Maximum stationarity error:",
        f"{stationarity_error:.3e}",
    )

    return {
        "divergence_matrix": D,
        "similarity_matrix": S,
        "tau": tau,
        "perron_eigenvalue": rho,
        "right_perron_vector": right,
        "left_perron_vector": left,
        "transition_matrix": P,
        "stationary_distribution": stationary,
    }


def plot_stationary_distribution(
    bandwidths: np.ndarray,
    stationary_distribution: np.ndarray,
) -> None:
    """Plot the stationary probability assigned to each bandwidth."""
    bandwidths = np.asarray(bandwidths, dtype=float)
    stationary_distribution = np.asarray(
        stationary_distribution,
        dtype=float,
    )

    critical_index = int(np.argmax(stationary_distribution))
    critical_bandwidth = bandwidths[critical_index]
    critical_probability = stationary_distribution[critical_index]

    plt.figure(figsize=(9, 5))

    plt.plot(
        bandwidths,
        stationary_distribution,
        marker="o",
        markersize=3,
        linewidth=1.5,
    )

    plt.axvline(
        critical_bandwidth,
        linestyle="--",
        label=(
            rf"Maximum at $h={critical_bandwidth:.4f}$"
        ),
    )

    plt.scatter(
        [critical_bandwidth],
        [critical_probability],
        s=70,
        zorder=3,
    )

    plt.xlabel(r"Bandwidth $h_i$")
    plt.ylabel(r"Stationary probability $\pi_i$")
    plt.title(
        "Stationary distribution of the Perron–Doob chain"
    )
    plt.legend()
    plt.tight_layout()
    plt.show()

    print("Selected index:", critical_index)
    print("Selected bandwidth:", critical_bandwidth)
    print("Stationary probability:", critical_probability)


# ============================================================
# Apply to the previously calculated KL matrix
# ============================================================

doob_result = doob_transform_from_divergence(
    divergence_matrix=kl_matrix_mc,
    tau=None,
)

stationary_distribution = doob_result[
    "stationary_distribution"
]

transition_matrix = doob_result["transition_matrix"]
similarity_matrix = doob_result["similarity_matrix"]
left_perron = doob_result["left_perron_vector"]
right_perron = doob_result["right_perron_vector"]

plot_stationary_distribution(
    bandwidths=bandwidths,
    stationary_distribution=stationary_distribution,
)
```

```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.special import logsumexp
from scipy.optimize import minimize_scalar
from scipy.stats import norm


# ============================================================
# Gaussian KDE evaluation
# ============================================================

def gaussian_kde_density(x, data, bandwidth):
    """Evaluate a one-dimensional Gaussian KDE."""
    x = np.asarray(x, dtype=float).reshape(-1)
    data = np.asarray(data, dtype=float).reshape(-1)

    if bandwidth <= 0:
        raise ValueError("bandwidth must be positive")

    z = (x[:, None] - data[None, :]) / bandwidth
    return np.mean(norm.pdf(z), axis=1) / bandwidth


def log_gaussian_kdes(
    evaluation_points,
    data,
    bandwidths,
    point_batch_size=500,
):
    """
    Evaluate log KDEs for several bandwidths.

    Returns an array with shape
        (number of evaluation points, number of bandwidths).
    """
    points = np.asarray(evaluation_points, dtype=float).reshape(-1)
    data = np.asarray(data, dtype=float).reshape(-1)
    bandwidths = np.asarray(bandwidths, dtype=float).reshape(-1)

    if np.any(bandwidths <= 0):
        raise ValueError("All bandwidths must be positive")

    n_points = len(points)
    n_bandwidths = len(bandwidths)
    n_data = len(data)

    result = np.empty((n_points, n_bandwidths))

    constants = (
        -np.log(n_data)
        - np.log(bandwidths)
        - 0.5 * np.log(2.0 * np.pi)
    )

    for start in range(0, n_points, point_batch_size):
        stop = min(start + point_batch_size, n_points)

        differences_squared = (
            points[start:stop, None] - data[None, :]
        ) ** 2

        for j, h in enumerate(bandwidths):
            log_kernel_values = (
                -differences_squared / (2.0 * h**2)
            )

            result[start:stop, j] = (
                constants[j]
                + logsumexp(log_kernel_values, axis=1)
            )

    return result


# ============================================================
# Monte Carlo KL matrix
# ============================================================

def gaussian_kde_kl_matrix_mc(
    data,
    bandwidths,
    n_mc=2000,
    random_state=None,
    show_progress=True,
):
    """
    Estimate

        D[i,j] = KL(f_hat_{h_i} || f_hat_{h_j})

    by Monte Carlo simulation.
    """
    data = np.asarray(data, dtype=float).reshape(-1)
    bandwidths = np.asarray(bandwidths, dtype=float).reshape(-1)

    rng = np.random.default_rng(random_state)

    n = len(data)
    number_of_bandwidths = len(bandwidths)

    kl_matrix = np.zeros(
        (number_of_bandwidths, number_of_bandwidths)
    )

    standard_errors = np.zeros_like(kl_matrix)

    # Common random numbers reduce simulation noise.
    component_indices = rng.integers(
        0,
        n,
        size=n_mc,
    )

    gaussian_noise = rng.standard_normal(n_mc)

    for i, h_source in enumerate(bandwidths):

        # Exact draws from the source KDE
        simulated_points = (
            data[component_indices]
            + h_source * gaussian_noise
        )

        log_densities = log_gaussian_kdes(
            evaluation_points=simulated_points,
            data=data,
            bandwidths=bandwidths,
        )

        log_source = log_densities[:, i]

        log_ratios = (
            log_source[:, None] - log_densities
        )

        kl_matrix[i, :] = np.mean(log_ratios, axis=0)

        standard_errors[i, :] = (
            np.std(log_ratios, axis=0, ddof=1)
            / np.sqrt(n_mc)
        )

        kl_matrix[i, i] = 0.0
        standard_errors[i, i] = 0.0

        if show_progress and (
            (i + 1) % 10 == 0
            or i == 0
            or i + 1 == number_of_bandwidths
        ):
            print(
                f"KL row {i + 1}/{number_of_bandwidths}"
            )

    return kl_matrix, standard_errors


# ============================================================
# Perron vectors and Doob transform
# ============================================================

def perron_eigenvector(matrix):
    """
    Return the Perron eigenvalue and a positive right eigenvector.

    The eigenvector is normalized to sum to one.
    """
    matrix = np.asarray(matrix, dtype=float)

    eigenvalues, eigenvectors = np.linalg.eig(matrix)

    perron_index = np.argmax(eigenvalues.real)

    eigenvalue = float(eigenvalues[perron_index].real)
    eigenvector = eigenvectors[:, perron_index].real

    # Resolve the arbitrary sign
    if np.sum(eigenvector) < 0:
        eigenvector = -eigenvector

    # Remove tiny numerical negative values
    eigenvector = np.maximum(eigenvector, 0.0)
    eigenvector /= np.sum(eigenvector)

    return eigenvalue, eigenvector


def doob_transform_from_divergence(
    divergence_matrix,
    tau=None,
):
    """
    Convert a divergence matrix D to

        S_ij = exp(-D_ij / tau),

    and compute the Perron–Doob Markov chain.
    """
    D = np.asarray(divergence_matrix, dtype=float).copy()

    if D.ndim != 2 or D.shape[0] != D.shape[1]:
        raise ValueError("D must be square")

    # Remove small negative Monte Carlo estimates
    D = np.maximum(D, 0.0)
    np.fill_diagonal(D, 0.0)

    n = D.shape[0]

    mask = ~np.eye(n, dtype=bool)
    positive_values = D[mask & (D > 0)]

    if tau is None:
        tau = float(np.median(positive_values))

    if tau <= 0:
        raise ValueError("tau must be positive")

    similarity_matrix = np.exp(-D / tau)

    # Right Perron vector
    rho_right, right = perron_eigenvector(
        similarity_matrix
    )

    # Left Perron vector
    rho_left, left = perron_eigenvector(
        similarity_matrix.T
    )

    # Bilinear Perron eigenvalue estimate
    rho = float(
        left @ similarity_matrix @ right
        / (left @ right)
    )

    # Doob transform
    transition_matrix = (
        similarity_matrix
        * right[None, :]
        / (rho * right[:, None])
    )

    # Natural combination of left and right Perron vectors
    stationary_distribution = left * right
    stationary_distribution /= stationary_distribution.sum()

    print(f"tau = {tau:.8f}")
    print(f"Perron eigenvalue = {rho:.8f}")

    print(
        "Maximum row-sum error:",
        np.max(
            np.abs(
                transition_matrix.sum(axis=1) - 1.0
            )
        ),
    )

    print(
        "Maximum stationarity error:",
        np.max(
            np.abs(
                stationary_distribution
                @ transition_matrix
                - stationary_distribution
            )
        ),
    )

    return {
        "divergence_matrix": D,
        "similarity_matrix": similarity_matrix,
        "transition_matrix": transition_matrix,
        "stationary_distribution": stationary_distribution,
        "right_perron_vector": right,
        "left_perron_vector": left,
        "perron_eigenvalue": rho,
        "tau": tau,
    }


def plot_stationary_distribution(
    bandwidths,
    stationary_distribution,
):
    """Plot the stationary Doob distribution."""
    bandwidths = np.asarray(bandwidths)
    stationary_distribution = np.asarray(
        stationary_distribution
    )

    selected_index = int(
        np.argmax(stationary_distribution)
    )

    selected_bandwidth = bandwidths[selected_index]

    plt.figure(figsize=(9, 5))

    plt.plot(
        bandwidths,
        stationary_distribution,
        marker="o",
        markersize=3,
    )

    plt.axvline(
        selected_bandwidth,
        linestyle="--",
        label=rf"Selected $h={selected_bandwidth:.4f}$",
    )

    plt.xlabel(r"Bandwidth $h$")
    plt.ylabel(r"Stationary probability $\pi(h)$")
    plt.title("Stationary distribution of the Doob chain")
    plt.legend()
    plt.tight_layout()
    plt.show()

    return selected_index, selected_bandwidth


# ============================================================
# Exact Gaussian MISE and oracle bandwidth
# ============================================================

def exact_gaussian_kde_mise(
    bandwidth,
    sample_size,
    sigma,
):
    """
    Exact MISE for Gaussian data and a Gaussian kernel.
    """
    h = float(bandwidth)

    if h <= 0:
        return np.inf

    term_1 = (
        1.0
        / (
            2.0
            * np.sqrt(np.pi)
            * sample_size
            * h
        )
    )

    term_2 = (
        (1.0 - 1.0 / sample_size)
        / (
            2.0
            * np.sqrt(np.pi)
            * np.sqrt(sigma**2 + h**2)
        )
    )

    term_3 = (
        -2.0
        / np.sqrt(
            2.0
            * np.pi
            * (2.0 * sigma**2 + h**2)
        )
    )

    term_4 = (
        1.0
        / (
            2.0
            * np.sqrt(np.pi)
            * sigma
        )
    )

    return term_1 + term_2 + term_3 + term_4


def compute_gaussian_mise_oracle(
    sample_size,
    sigma,
    lower_bound=1e-5,
    upper_bound=5.0,
):
    """Find the exact continuous MISE-optimal bandwidth."""
    result = minimize_scalar(
        lambda h: exact_gaussian_kde_mise(
            h,
            sample_size,
            sigma,
        ),
        bounds=(lower_bound, upper_bound),
        method="bounded",
    )

    if not result.success:
        raise RuntimeError(result.message)

    return float(result.x), float(result.fun)


# ============================================================
# Final comparison plot
# ============================================================

def plot_kde_comparison(
    data,
    sigma,
    h_spectral,
    h_oracle,
):
    """
    Plot the true density, spectral KDE, and oracle KDE.
    """
    data = np.asarray(data)

    largest_h = max(h_spectral, h_oracle)

    x_min = min(data.min(), -5.0 * sigma) - 3 * largest_h
    x_max = max(data.max(), 5.0 * sigma) + 3 * largest_h

    x_grid = np.linspace(x_min, x_max, 4000)

    true_density = norm.pdf(
        x_grid,
        loc=0.0,
        scale=sigma,
    )

    spectral_density = gaussian_kde_density(
        x_grid,
        data,
        h_spectral,
    )

    oracle_density = gaussian_kde_density(
        x_grid,
        data,
        h_oracle,
    )

    spectral_ise = np.trapz(
        (spectral_density - true_density) ** 2,
        x=x_grid,
    )

    oracle_ise = np.trapz(
        (oracle_density - true_density) ** 2,
        x=x_grid,
    )

    plt.figure(figsize=(10, 6))

    plt.plot(
        x_grid,
        true_density,
        linewidth=2.5,
        label=rf"True $N(0,{sigma}^2)$",
    )

    plt.plot(
        x_grid,
        spectral_density,
        linewidth=2,
        label=rf"KL spectral KDE: $h={h_spectral:.4f}$",
    )

    plt.plot(
        x_grid,
        oracle_density,
        linestyle="--",
        linewidth=2,
        label=rf"MISE oracle KDE: $h={h_oracle:.4f}$",
    )

    plt.vlines(
        data,
        ymin=0,
        ymax=0.01 * true_density.max(),
        alpha=0.25,
        linewidth=0.5,
    )

    plt.xlabel(r"$x$")
    plt.ylabel("Density")
    plt.title("KL-spectral bandwidth versus MISE oracle")
    plt.legend()
    plt.tight_layout()
    plt.show()

    print(f"Observed spectral ISE: {spectral_ise:.10f}")
    print(f"Observed oracle ISE:   {oracle_ise:.10f}")
```

```python
import numpy as np


def trapezoid(y, x=None, dx=1.0):
    """
    Approximate the integral of sampled values y using the
    composite trapezoidal rule.

    Parameters
    ----------
    y : array-like
        Function values y_0, ..., y_m.
    x : array-like or None
        Corresponding grid points. They need not be equally spaced.
        If x is None, an equally spaced grid with spacing dx is used.
    dx : float
        Grid spacing when x is None.

    Returns
    -------
    float
        Approximation of the integral.
    """
    y = np.asarray(y, dtype=float)

    if y.ndim != 1:
        raise ValueError("y must be one-dimensional.")

    if y.size < 2:
        raise ValueError("At least two function values are required.")

    if x is None:
        return float(
            dx * np.sum((y[:-1] + y[1:]) / 2.0)
        )

    x = np.asarray(x, dtype=float)

    if x.ndim != 1:
        raise ValueError("x must be one-dimensional.")

    if x.size != y.size:
        raise ValueError("x and y must have the same length.")

    return float(
        np.sum(
            (x[1:] - x[:-1])
            * (y[:-1] + y[1:])
            / 2.0
        )
    )
```

```python
def doob_transform_from_divergence_safe(
    divergence_matrix,
    tau,
    remove_diagonal=False,
    min_similarity=1e-300,
):
    D = np.asarray(divergence_matrix, dtype=float).copy()

    if D.ndim != 2 or D.shape[0] != D.shape[1]:
        raise ValueError("D must be square.")

    if tau <= 0:
        raise ValueError("tau must be strictly positive.")

    D = np.maximum(D, 0.0)
    np.fill_diagonal(D, 0.0)

    # Prevent exponential underflow.
    exponent = np.clip(
        -D / tau,
        -700.0,
        0.0,
    )

    S = np.exp(exponent)
    S = np.maximum(S, min_similarity)

    if remove_diagonal:
        np.fill_diagonal(S, 0.0)

        # Add a tiny amount everywhere to preserve positivity.
        S += min_similarity

    # Positive power iteration
    def power_iteration(A, tol=1e-13, max_iter=100_000):
        n = A.shape[0]
        v = np.full(n, 1.0 / n)

        for _ in range(max_iter):
            v_new = A @ v
            norm = np.sum(v_new)

            if not np.isfinite(norm) or norm <= 0:
                raise RuntimeError(
                    "Power iteration failed. Tau may be too small."
                )

            v_new /= norm

            if np.linalg.norm(v_new - v, ord=1) < tol:
                v = v_new
                break

            v = v_new

        Av = A @ v
        rho = float(
            np.dot(v, Av) / np.dot(v, v)
        )

        return rho, v

    rho_right, right = power_iteration(S)
    rho_left, left = power_iteration(S.T)

    # Avoid division by numerical zeros
    tiny = np.finfo(float).tiny
    right = np.maximum(right, tiny)
    left = np.maximum(left, tiny)

    right /= right.sum()
    left /= left.sum()

    rho = float(
        left @ (S @ right)
        / (left @ right)
    )

    P = (
        S
        * right[None, :]
        / (rho * right[:, None])
    )

    # Correct small floating-point row-sum errors
    P /= P.sum(axis=1, keepdims=True)

    pi = left * right
    pi /= pi.sum()

    return {
        "similarity_matrix": S,
        "transition_matrix": P,
        "stationary_distribution": pi,
        "right_perron_vector": right,
        "left_perron_vector": left,
        "perron_eigenvalue": rho,
        "tau": tau,
        "minimum_off_diagonal_similarity": np.min(
            S[~np.eye(S.shape[0], dtype=bool)]
        ),
    }
```

```python
tau_values = [
    0.001,
    0.005,
    0.01,
    0.02,
    0.05,
    0.1,
    0.2,
    0.5,
    1.0,
]

selected_bandwidths = []

plt.figure(figsize=(10, 6))

for tau in tau_values:
    try:
        result = doob_transform_from_divergence_safe(
            divergence_matrix=kl_matrix_mc,
            tau=tau,
            remove_diagonal=False,
        )

        pi = result["stationary_distribution"]

        selected_index = int(np.argmax(pi))
        selected_h = float(bandwidths[selected_index])

        selected_bandwidths.append(selected_h)

        minimum_similarity = result[
            "minimum_off_diagonal_similarity"
        ]

        print(
            f"tau={tau:.4g}, "
            f"selected h={selected_h:.6f}, "
            f"minimum off-diagonal similarity="
            f"{minimum_similarity:.3e}"
        )

        plt.plot(
            bandwidths,
            pi,
            label=rf"$\tau={tau}$",
        )

    except (RuntimeError, FloatingPointError) as error:
        selected_bandwidths.append(np.nan)

        print(
            f"tau={tau:.4g} skipped: {error}"
        )

plt.axvline(
    h_oracle,
    linestyle="--",
    linewidth=2,
    label=rf"MISE oracle $h={h_oracle:.4f}$",
)

plt.xlabel(r"Bandwidth $h$")
plt.ylabel(r"Stationary probability $\pi_i$")
plt.title("Stationary distributions for different tau values")
plt.legend(ncol=2)
plt.tight_layout()
plt.show()
```
