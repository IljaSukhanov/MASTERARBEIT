# Jones AR(2) coefficient sample

Run the cells in order to draw and plot the coefficients. The shared [plotting helper](/Users/iljasukhanov/MASTERARBEIT/Code/Code_AR_processes/ar_tools/plotting_functions.py) saves the coefficient CSV, a PNG, and a PDF under the corresponding `outputs` folders.


```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import Image, display

AR_DIR = next(
    candidate.resolve()
    for root in (Path.cwd(), *Path.cwd().parents)
    for candidate in (root / "Code" / "Code_AR_processes", root)
    if (candidate / "ar_tools/experiments.py").is_file()
)
PROJECT_ROOT = AR_DIR.parents[1]
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

with plt.rc_context():
    from Code.Code_AR_processes.ar_tools.plotting_functions import plot_ar2_jones_coefficient_sample

EXPERIMENT_NAME = "ar2_jones_coefficient_sample"
SEED = 42
SAMPLE_COUNT = 500
```

```python
with plt.rc_context():
    plot_ar2_jones_coefficient_sample(
        ar_directory=AR_DIR, seed=SEED, sample_count=SAMPLE_COUNT,
        name=EXPERIMENT_NAME,
    )
```

```python
csv_path = AR_DIR / "outputs/csv" / EXPERIMENT_NAME / "coefficients.csv"
image_path = AR_DIR / "outputs/images" / EXPERIMENT_NAME / f"{EXPERIMENT_NAME}.png"
pdf_path = AR_DIR / "outputs/pdfs" / EXPERIMENT_NAME / f"{EXPERIMENT_NAME}.pdf"
coefficients = pd.read_csv(csv_path)
assert len(coefficients) == SAMPLE_COUNT
display(coefficients.head())
display(Image(filename=str(image_path)))
print(f"PDF: {pdf_path}")
```
